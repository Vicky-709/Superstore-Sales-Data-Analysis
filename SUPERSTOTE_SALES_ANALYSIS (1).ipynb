## DATA ANALYSIS

```python
# TOTAL SALES
total_sales = df["Sales"].sum()
print("Total Sales:",total_sales)
```

```python
# TOTAL PROFIT
total_profit = df["Profit"].sum()
print("Total Profit:",total_profit)
```

```python
# SALES BY CATEGORY
category_profit = df.groupby("Category")["Sales"].sum()
category_profit
```

```python
# TOP 10 PRODUCTS BY SALES
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
top_products
```

```python
# SALES BY REGION
region_sales = df.groupby("Region")["Sales"].sum()
print(region_sales.sort_values(ascending=False))
```

```python
# MONTHLY SALES
df["Order Date"] = pd.to_datetime(df["Order Date"])
monthly_sales = df.groupby(df["Order Date"].dt.month)["Sales"].sum()
monthly_sales
```

## BUSINESS INSIGHTS

```python
# SALES AND PROFIT BY SEGMENT
segment_analysis = df.groupby("Segment").agg({"Sales": "sum","Profit": "sum"})
print(segment_analysis.sort_values(by="Sales",ascending=False))
```

```python
# DISCOUNT IMPACT ON PROFIT
discount_analysis = df.groupby("Discount").agg({"Sales": "sum","Profit": "sum"})
discount_analysis
```

```python
# TOP 5 PROFITABLE SUB-CATEGORY
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False).head(5)
subcategory_profit
```

```python
# SALES BY SHIPPING MODE
shipping_analysis = df.groupby("Ship Mode").agg({"Sales": "sum","Profit": "sum","Order ID":"nunique"})
print(shipping_analysis.sort_values(by="Sales",ascending=False))
```

```python
# TOP 10 CUSTOMERS
top_customers = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(10)
top_customers
```

```python
# LOSS MAKING SUB-CATEGORY
loss_categories = subcategory_profit[subcategory_profit < 0]
loss_categories
```

```python
# PROFIT MARGIN BY CATEGORY
category_summary = df.groupby("Category").agg({"Sales": "sum","Profit": "sum"})
category_summary["Profit Margin (%)"] = (category_summary["Profit"] /category_summary["Sales"] * 100)
category_summary
```

```python
# NEGATIVE PROFIT TRANSACTIONS
loss_records = df[df["Profit"] < 0]

print("Loss-making records:", len(loss_records))

print(loss_records[[
    "Order ID",
    "Product Name",
    "Sales",
    "Discount",
    "Profit"
]].head(10))
```

## 1. Key Business Insights

1. Technology is the highest-performing category in terms of sales and profit.
2. Furniture has high sales but a comparatively low profit margin.
3. The West region generates the highest sales and profit.
4. Copiers generate the highest profit among all sub-categories.
5. Tables, Bookcases and Supplies have negative overall profits.
6. Higher discount levels are associated with negative profitability.
7. Sean Miller is the highest-sales customer.

## 2. Business Recommendations

1. Focus on expanding profitable Technology products.
2. Review pricing and discount strategies for Furniture.
3. Investigate losses in Tables, Bookcases and Supplies.
